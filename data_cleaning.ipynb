```python
# load the csv data 
data = pd.read_csv("dataset2.csv", encoding='latin-1')
```

```python
import pandas as pd

# Define the 20 most impactful columns for analysis
meaningful_cols = [
    # 1. Time & Logistics
    'eventid', 'iyear', 'imonth', 'iday', 'extended',
    
    # 2. Location Analysis
    'country_txt', 'region_txt', 'provstate', 'city', 'latitude', 'longitude',
    
    # 3. The Tactics & Targets
    'attacktype1_txt', 'targtype1_txt', 'targsubtype1_txt', 'gname', 'weaptype1_txt',
    
    # 4. Casualties & Impact Assessment
    'nkill', 'nwound', 'property'
]

# Create a clean, focused dataframe
df_clean = data[meaningful_cols].copy()

# Double check your new lightweight dataframe shape
print("New Shape:", df_clean.shape)
df_clean.info()
```

```python
# 1. Handle Numerical Missing Values (Casualties)
# Replacing missing deaths and injuries with 0 so sum() and mean() functions work perfectly
df_clean['nkill'] = df_clean['nkill'].fillna(0)
df_clean['nwound'] = df_clean['nwound'].fillna(0)

# 2. Handle Text/String Missing Values
# Replacing missing locations and sub-targets with a placeholder string 'Unknown'
string_cols_to_fill = ['provstate', 'city', 'targsubtype1_txt']
df_clean[string_cols_to_fill] = df_clean[string_cols_to_fill].fillna('Unknown')


df_clean['total_casualties'] = df_clean['nkill'] + df_clean['nwound']

# Verification check to see if any NaNs remain (except latitude/longitude)
print("Remaining Missing Values per Column:")


# drop latitude and longitude missing rows 
df_clean = df_clean.dropna(subset=['latitude', 'longitude'])
```

```python
df_clean.to_csv("final_dataset.csv")
```
